```python
#importar bibliotecas
from pymongo import MongoClient
import pandas as pd
import json

# Conecte-se ao servidor MongoDB
client = MongoClient('mongodb+srv://STRING_MONGODB')
```

###### **Consulta e Parâmetros**

Consultar collection Vagas e configurar para o consumo desses dados

```python
# Buscar a collection e banco de dados
banco_dados = client.BASE_DE_DADOS
collection_vagas = client.BASE_DE_DADOS.vagas # collection vagas
collection_users = client.BASE_DE_DADOS.users # collection users

# Consulta para encontrar todos os documentos na coleção
cursor_vagas = collection_vagas.find()# cursos vagas
cursor_users = collection_users.find() # cursos users
output_path = "DIRETÓRIO QUE SERÁ SALVO"
```

**Extração de dados da collection VAGAS**

Selecionar a collection vagas e retornar o exporta no formato de JSON 

```python
# configuração para consumo dos dados
def exportar_para_json(banco_dados, collection_vagas, output_path, criar_diretorio=True):

    # Converter os documentos para uma lista de dicionários
    documentos_lista = list(cursor_vagas)


    # Especificar o caminho do arquivo JSON de saída
    with open(output_path, 'w') as arquivo_json:
        # Escrever a lista de documentos em um arquivo JSON
        json.dump(documentos_lista, arquivo_json, default=str)
# Substitua 'seu_banco_de_dados', 'sua_colecao' e '/caminho/para/saida/dados.json' pelos valores desejados.
exportar_para_json(banco_dados, collection_vagas, output_path + 'Vagas.json', criar_diretorio=True)
```
